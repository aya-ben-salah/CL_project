## Visualisation de stress_level

```python
c1= df["stress_level"].value_counts()
plt.bar(c1.index , c1.values , color="black")
plt.title("bar_stress_level")
plt.xlabel("stress_level")
plt.ylabel("Fréquence")

plt.show()
```

```python
# Compter les occurrences de chaque niveau de stress
counts = df["stress_level"].value_counts()

# Créer le camembert
plt.pie(counts, labels=counts.index, autopct="%0.1f%%")
plt.title("Répartition du stress")
plt.show()
```

```python

# 1. Compter les occurrences et trier par fréquence décroissante
counts = df["stress_level"].value_counts().sort_values(ascending=False)

# 2. Calculer le pourcentage cumulatif
cumulative_percent = counts.cumsum() / counts.sum() * 100

# 3. Créer le diagramme
fig, ax1 = plt.subplots()

# Histogramme (barres)
ax1.bar(counts.index, counts, color='skyblue')
ax1.set_ylabel('Fréquence')
ax1.set_xlabel('Niveau de stress')

# Courbe cumulative (%)
ax2 = ax1.twinx()  # créer un deuxième axe y
ax2.plot(counts.index, cumulative_percent, color='red', marker='o', linestyle='-', linewidth=2)
ax2.set_ylabel('Pourcentage cumulatif')
ax2.set_ylim(0, 110)  # un peu au-dessus de 100 pour que le sommet ne touche pas le cadre

# Ajouter un titre
plt.title('Diagramme de Pareto du stress')

plt.show()
```

- Les trois catégories (Low, Medium, High) présentent des fréquences similaires.
- La distribution est donc globalement équilibrée entre les différents niveaux de stress.
- Cela montre qu’il n’y a pas de niveau de stress dominant dans la population étudiée.
- Le stress est réparti de manière homogène, ce qui indique une diversité des situations individuelles.

Asymétrie : Les distributions sont effectivement asymétriques, avec une concentration plus haute pour les niveaux "Moderate" et "Severe".

Représentation : Les catégories "Moderate" et "Severe" sont les plus denses (elles contiennent plus de points), tandis que la catégorie "Mild" est la moins représentée.

Écart inter-catégories : On observe une différence nette entre "Mild" et "Moderate". En revanche, les distributions de "Moderate" et "Severe" sont presque identiques (différence négligeable).

Variabilité élevée : Contrairement à l'apparence "serrée" des points, la variabilité est très forte pour "Moderate" et "Severe" car les points s'étalent sur une plage très large (de 3h à 12h).

```python
sns.barplot(x="addiction_level", y="daily_screen_time_hours", data=df)
plt.title ("barplot de daily_screen_time_hours avec addiction_level")
plt.xlabel("addiction_lable")
plt.ylabel("Moyenne de daily_screen_time_hours par addiction_level")

plt.show()
```

Différence marquée : L'écart de moyenne est important entre "Mild" et "Moderate", mais presque négligeable entre "Moderate" et "Severe".

Domination : Les niveaux d'addiction "Moderate" et "Severe" présentent les consommations les plus fortes.

Précision : Les barres d'erreur très courtes indiquent une estimation très précise de la moyenne pour chaque catégorie.
