# variable qualitative _ quantitative ( country _ video_views)

```python
df=pd.read_csv(r"C:\Users\hp\Desktop\project (1)\data\Global-YouTube-Statistics.csv",encoding = "ISO-8859-1")
df.head()
```

```python
#Top 3 countries 
top_3_countries = df["Country"].value_counts().head(3).index
top_3_countries
```

# Justification
Nous avons choisi de conserver uniquement les 3 pays les plus représentés dans le dataset.

```python
#boxplot
plt.figure(figsize=(12, 6))
sns.boxplot(x="Country",y="video views",data=filtered_df)
plt.title("boxplot Country vs vido_views")
plt.show()
```

# Interpretation
Le boxplot met en évidence une forte variabilité du nombre de vues selon les pays étudiés. Bien que les médianes soient relativement proches, l’Inde et les États-Unis se distinguent par la présence de nombreuses valeurs aberrantes, traduisant l’existence de vidéos ayant généré un volume de vues exceptionnellement élevé. À l’inverse, la catégorie “Unknown” présente une dispersion plus faible et des valeurs extrêmes moins marquées

<h3>countplot croise</h3>

```python
import seaborn as sns
import matplotlib.pyplot as plt

top10 = df["Country"].value_counts().head(10).index

sns.countplot(data=df[df["Country"].isin(top10)], x="Country", hue="channel_type")
plt.title("Country vs channel_type")
plt.xticks(rotation=45)
plt.show()
```

interpr!tqtion /

Relation entre les variables : Il existe une relation entre le pays et le type de chaîne, car la distribution des catégories varie selon les pays.

Différences entre pays : India et United States ont les nombres de chaînes les plus élevés.

Catégories dominantes : Les catégories Music et Entertainment sont parmi les plus fréquentes.

Combinaisons rares : Certaines catégories comme Autos, Nonprofit ou Animals apparaissent très peu.

Déséquilibre : Il existe un déséquilibre structurel, car la majorité des chaînes est concentrée dans quelques pays.

<h3>diargame en barre regrouper<h3>

```python
import pandas as pd
import matplotlib.pyplot as plt

top10 = df["Country"].value_counts().head(10).index

table = pd.crosstab(
    df[df["Country"].isin(top10)]["Country"],
    df[df["Country"].isin(top10)]["channel_type"]
)

table.plot(kind="bar", figsize=(10,5))

plt.title("Country vs Channel Type")
plt.xlabel("Country")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

<h3>diagramme en berres empilées</h3>

```python
import pandas as pd
import matplotlib.pyplot as plt

top10 = df["Country"].value_counts().head(10).index

table = pd.crosstab(
    df[df["Country"].isin(top10)]["Country"],
    df[df["Country"].isin(top10)]["channel_type"]
)

table.plot(kind="bar", stacked=True, figsize=(10,5))

plt.title("Country vs Channel Type")
plt.xlabel("Country")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

interpretation

Relation entre variables : Il existe une relation entre le pays et le type de chaîne, car certains types apparaissent plus dans certains pays.

Pays dominants : United States et India ont les nombres de chaînes les plus élevés.

Catégories dominantes : Entertainment et Music sont les types de chaînes les plus fréquents.

Combinaisons rares : Certaines catégories comme Animals, Nonprofit ou Autos apparaissent très peu dans plusieurs pays.

Déséquilibre : La majorité des chaînes est concentrée dans quelques pays, surtout United States et India.

<h3>Heatmap du tableau de contingence</h3>

```python
import seaborn as sns
import matplotlib.pyplot as plt

top10 = df["Country"].value_counts().head(10).index

table = pd.crosstab(
    df[df["Country"].isin(top10)]["Country"],
    df[df["Country"].isin(top10)]["channel_type"]
)

plt.figure(figsize=(10,6))
sns.heatmap(table, annot=True, cmap="YlGnBu")

plt.title("Heatmap: Country vs Channel Type")
plt.xlabel("Channel Type")
plt.ylabel("Country")
plt.show()
```

inter

Les États-Unis dominent largement, surtout dans Entertainment et Music.

L’Inde arrive en deuxième, principalement dans Entertainment et Music.

Entertainment est la catégorie la plus dominante dans presque tous les pays.

Plusieurs catégories comme Animals, Autos, Nonprofit et Tech sont très peu représentées.

<h3>décision<h3>

Se concentrer sur les catégories Entertainment et Music dans les pays United States et India.


Ces pays ont le plus grand nombre de chaînes, et ces catégories sont les plus dominantes. Se focaliser sur ces combinaisons permet donc de viser le marché le plus actif et le plus populaire.
