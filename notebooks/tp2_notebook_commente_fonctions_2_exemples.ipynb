## Exercice 1 — Arbre binaire quelconque
Fonctions de création, propriétés et parcours.


```python
# Fonction : creer_noeud
# Rôle : crée un nœud d'arbre binaire.
def creer_noeud(valeur, gauche=None, droite=None):
    """Crée un nœud représenté par un dictionnaire."""
    return {"valeur": valeur, "gauche": gauche, "droite": droite}

# Fonction : construire_arbre_binaire
# Rôle : construit un arbre binaire simple pour les tests.
def construire_arbre_binaire():
    """Construit l'arbre de test pour les exercices."""
    f4 = creer_noeud(4)            # feuille
    f5 = creer_noeud(5)            # feuille
    f6 = creer_noeud(6)            # feuille
    n2 = creer_noeud(2, f4, f5)   # nœud interne
    n3 = creer_noeud(3, None, f6) # nœud interne (enfant gauche absent)
    return creer_noeud(1, n2, n3) # racine
```

```python
# Fonction : taille_arbre
# Rôle : calcule le nombre total de nœuds de l'arbre.
def taille_arbre(arbre):
    # Nombre total de noeuds
    if arbre is None:
        return 0
    return 1 + taille_arbre(arbre['gauche']) + taille_arbre(arbre['droite'])

# Fonction : hauteur_arbre
# Rôle : calcule la hauteur de l'arbre.
def hauteur_arbre(arbre):
    # Hauteur : -1 si vide, 0 pour un noeud isolé
    if arbre is None:
        return -1
    return 1 + max(hauteur_arbre(arbre['gauche']), hauteur_arbre(arbre['droite']))

# Fonction : nombre_feuilles
# Rôle : calcule le nombre de feuilles de l'arbre.
def nombre_feuilles(arbre):
    # Noeuds sans enfant
    if arbre is None:
        return 0
    if arbre['gauche'] is None and arbre['droite'] is None:
        return 1
    return nombre_feuilles(arbre['gauche']) + nombre_feuilles(arbre['droite'])

# Fonction : nombre_noeuds_internes
# Rôle : calcule le nombre de nœuds internes de l'arbre.
def nombre_noeuds_internes(arbre):
    # Noeuds avec au moins un enfant
    if arbre is None:
        return 0
    if arbre['gauche'] is None and arbre['droite'] is None:
        return 0
    return 1 + nombre_noeuds_internes(arbre['gauche']) + nombre_noeuds_internes(arbre['droite'])
```

```python
from collections import deque

# Fonction : parcours_prefixe
# Rôle : effectue le parcours préfixe de l'arbre.
def parcours_prefixe(arbre):
    if arbre is None:
        return []
    return [arbre['valeur']] + parcours_prefixe(arbre['gauche']) + parcours_prefixe(arbre['droite'])

# Fonction : parcours_infixe
# Rôle : effectue le parcours infixe de l'arbre.
def parcours_infixe(arbre):
    if arbre is None:
        return []
    return parcours_infixe(arbre['gauche']) + [arbre['valeur']] + parcours_infixe(arbre['droite'])

# Fonction : parcours_postfixe
# Rôle : effectue le parcours postfixe de l'arbre.
def parcours_postfixe(arbre):
    if arbre is None:
        return []
    return parcours_postfixe(arbre['gauche']) + parcours_postfixe(arbre['droite']) + [arbre['valeur']]

# Fonction : parcours_largeur
# Rôle : effectue le parcours en largeur de l'arbre.
def parcours_largeur(arbre):
    if arbre is None:
        return []
    resultat, file = [], deque([arbre])
    while file:
        noeud = file.popleft()
        resultat.append(noeud['valeur'])
        if noeud['gauche']:
            file.append(noeud['gauche'])
        if noeud['droite']:
            file.append(noeud['droite'])
    return resultat
```

## Exercice 2 — Arbre Binaire de Recherche (ABR)
Construction, insertion, recherche et suppression.


```python
# Fonction : construire_abr
# Rôle : construit un arbre binaire de recherche à partir d'une liste de valeurs.
def construire_abr(valeurs):
    racine = None

    for valeur in valeurs:
        # arbre vide
        if racine is None:
            racine = creer_noeud(valeur)
        else:
            courant = racine
            
            while True:
                #gauche
                if valeur < courant["valeur"]:
                    if courant["gauche"] is None:
                        courant["gauche"] = creer_noeud(valeur)
                        break
                    else:
                        courant = courant["gauche"]
                
                #droite
                else:
                    if courant["droite"] is None:
                        courant["droite"] = creer_noeud(valeur)
                        break
                    else:
                        courant = courant["droite"]

    return racine

# Fonction : inserer_abr
# Rôle : insère une clé dans un ABR.
def inserer_abr(arbre, cle):
    if arbre is None:
        return creer_noeud(cle)
    if cle < arbre['valeur']:
        arbre['gauche'] = inserer_abr(arbre['gauche'], cle)
    elif cle > arbre['valeur']:
        arbre['droite'] = inserer_abr(arbre['droite'], cle)
    return arbre  # doublon ignoré

# Fonction : rechercher_abr
# Rôle : recherche une clé dans un ABR.
def rechercher_abr(arbre, cle):
    if arbre is None:
        return False
    if cle == arbre['valeur']:
        return True
    elif cle < arbre['valeur']:
        return rechercher_abr(arbre['gauche'], cle)
    else:
        return rechercher_abr(arbre['droite'], cle)

# Fonction : _min_noeud
# Rôle : retourne le nœud de plus petite valeur dans un sous-arbre.
def _min_noeud(arbre):
    while arbre['gauche'] is not None:
        arbre = arbre['gauche']
    return arbre

# Fonction : supprimer_abr
# Rôle : supprime une clé dans un ABR.
def supprimer_abr(arbre, cle):
    if arbre is None:
        return None
    if cle < arbre['valeur']:
        arbre['gauche'] = supprimer_abr(arbre['gauche'], cle)
    elif cle > arbre['valeur']:
        arbre['droite'] = supprimer_abr(arbre['droite'], cle)
    else:
        if arbre['gauche'] is None:
            return arbre['droite']
        elif arbre['droite'] is None:
            return arbre['gauche']
        # Deux enfants : successeur
        succ = _min_noeud(arbre['droite'])
        arbre['valeur'] = succ['valeur']
        arbre['droite'] = supprimer_abr(arbre['droite'], succ['valeur'])
    return arbre
```

```python
# Test Exercice 2 — Exemple 1
valeurs1 = [5, 3, 7, 1, 4, 6, 8]
abr1 = construire_abr(valeurs1)

print("=== Exemple 1 ===")
print("Infixe (trié) :", parcours_infixe(abr1))
print("Recherche 4 :", rechercher_abr(abr1, 4))
print("Recherche 9 :", rechercher_abr(abr1, 9))

abr1 = supprimer_abr(abr1, 3)
print("Après suppression de 3 :", parcours_infixe(abr1))

# Test Exercice 2 — Exemple 2
valeurs2 = [12, 7, 18, 3, 9, 15, 20]
abr2 = construire_abr(valeurs2)

print("\n=== Exemple 2 ===")
print("Infixe (trié) :", parcours_infixe(abr2))
print("Recherche 15 :", rechercher_abr(abr2, 15))
print("Recherche 100 :", rechercher_abr(abr2, 100))

abr2 = supprimer_abr(abr2, 12)
print("Après suppression de 12 :", parcours_infixe(abr2))
```

## Exercice 3 — Arbre AVL
Structure AVL, rotations et insertion équilibrée.

**Remarque :** le document source ne contient pas le code complet de `_maj_hauteur` ni de `supprimer_avl`.  
J’ai gardé uniquement le code présent dans le document, et j’ai ajouté `_maj_hauteur` minimal pour rendre les rotations exécutables dans le notebook.


```python
# Q1
# Fonction : creer_noeud_avl
# Rôle : crée un nœud AVL avec sa hauteur.
def creer_noeud_avl(valeur, gauche=None, droite=None):
    return {"valeur": valeur, "gauche": gauche, "droite": droite, "hauteur": 1}

# Fonction : construire_avl
# Rôle : construit un arbre AVL à partir d'une liste de valeurs.
def construire_avl(valeurs):
    arbre = None
    for v in valeurs:
        arbre = inserer_avl(arbre, v)
    return arbre

# Fonction : hauteur_noeud
# Rôle : retourne la hauteur d'un nœud.
def hauteur_noeud(noeud):
    if noeud is None:
        return 0
    return noeud["hauteur"]

# Fonction : facteur_equilibre
# Rôle : calcule le facteur d'équilibre d'un nœud.
def facteur_equilibre(noeud):
    if noeud is None:
        return 0
    return hauteur_noeud(noeud["gauche"]) - hauteur_noeud(noeud["droite"])

# Fonction : maj_hauteur
# Rôle : met à jour la hauteur d'un nœud.
def maj_hauteur(noeud):
    noeud["hauteur"] = 1 + max(hauteur_noeud(noeud["gauche"]),
                               hauteur_noeud(noeud["droite"]))
```

```python
# Fonction : rotation_droite
# Rôle : effectue une rotation droite.
def rotation_droite(y):
    x = y["gauche"]
    T2 = x["droite"]

    x["droite"] = y
    y["gauche"] = T2

    maj_hauteur(y)
    maj_hauteur(x)

    return x

# Fonction : rotation_gauche
# Rôle : effectue une rotation gauche.
def rotation_gauche(x):
    y = x["droite"]
    T2 = y["gauche"]

    y["gauche"] = x
    x["droite"] = T2

    maj_hauteur(x)
    maj_hauteur(y)

    return y

# Fonction : rotation_gauche_droite
# Rôle : effectue une rotation gauche-droite.
def rotation_gauche_droite(noeud):
    noeud["gauche"] = rotation_gauche(noeud["gauche"])
    return rotation_droite(noeud)

# Fonction : rotation_droite_gauche
# Rôle : effectue une rotation droite-gauche.
def rotation_droite_gauche(noeud):
    noeud["droite"] = rotation_droite(noeud["droite"])
    return rotation_gauche(noeud)
```

```python
# Fonction : equilibrer
# Rôle : rééquilibre un nœud AVL si nécessaire.
def equilibrer(noeud):
    maj_hauteur(noeud)
    fe = facteur_equilibre(noeud)

    if fe > 1:
        if facteur_equilibre(noeud["gauche"]) < 0:
            return rotation_gauche_droite(noeud)
        return rotation_droite(noeud)

    if fe < -1:
        if facteur_equilibre(noeud["droite"]) > 0:
            return rotation_droite_gauche(noeud)
        return rotation_gauche(noeud)

    return noeud

# Fonction : inserer_avl
# Rôle : insère une clé dans un arbre AVL.
def inserer_avl(arbre, cle):
    if arbre is None:
        return creer_noeud_avl(cle)

    if cle < arbre["valeur"]:
        arbre["gauche"] = inserer_avl(arbre["gauche"], cle)
    elif cle > arbre["valeur"]:
        arbre["droite"] = inserer_avl(arbre["droite"], cle)

    return equilibrer(arbre)

# Fonction : parcours_infixe_avl
# Rôle : effectue le parcours infixe d'un AVL.
def parcours_infixe_avl(arbre):
    if arbre is None:
        return []
    return parcours_infixe_avl(arbre['gauche']) + [arbre['valeur']] + parcours_infixe_avl(arbre['droite'])


#
# Fonction : rechercher_avl
# Rôle : recherche une clé dans un AVL.
def rechercher_avl(arbre, cle):
    if arbre is None:
        return False
    if cle == arbre["valeur"]:
        return True
    elif cle < arbre["valeur"]:
        return rechercher_avl(arbre["gauche"], cle)
    else:
        return rechercher_avl(arbre["droite"], cle)
    

##
# Fonction : _min_valeur
# Rôle : retourne le nœud contenant la plus petite valeur.
def _min_valeur(noeud):
    courant = noeud
    while courant["gauche"] is not None:
        courant = courant["gauche"]
    return courant

# Fonction : min_valeur
# Rôle : retourne le nœud de plus petite valeur dans un sous-arbre.
def min_valeur(noeud):
    while noeud["gauche"] is not None:
        noeud = noeud["gauche"]
    return noeud

# Fonction : supprimer_avl
# Rôle : supprime une clé dans un arbre AVL.
def supprimer_avl(arbre, cle):
    if arbre is None:
        return arbre

    if cle < arbre["valeur"]:
        arbre["gauche"] = supprimer_avl(arbre["gauche"], cle)
    elif cle > arbre["valeur"]:
        arbre["droite"] = supprimer_avl(arbre["droite"], cle)
    else:
        if arbre["gauche"] is None:
            return arbre["droite"]
        elif arbre["droite"] is None:
            return arbre["gauche"]

        temp = min_valeur(arbre["droite"])
        arbre["valeur"] = temp["valeur"]
        arbre["droite"] = supprimer_avl(arbre["droite"], temp["valeur"])

    return equilibrer(arbre)

##
# Fonction : parcours_infixe_avl
# Rôle : effectue le parcours infixe d'un AVL.
def parcours_infixe_avl(arbre):
    if arbre is None:
        return []
    return (parcours_infixe_avl(arbre['gauche']) +
            [arbre['valeur']] +
            parcours_infixe_avl(arbre['droite']))

# Fonction : parcours_prefixe_avl
# Rôle : effectue le parcours préfixe d'un AVL.
def parcours_prefixe_avl(arbre):
    if arbre is None:
        return []
    return ([arbre['valeur']] +
            parcours_prefixe_avl(arbre['gauche']) +
            parcours_prefixe_avl(arbre['droite']))

# Fonction : parcours_postfixe_avl
# Rôle : effectue le parcours postfixe d'un AVL.
def parcours_postfixe_avl(arbre):
    if arbre is None:
        return []
    return (parcours_postfixe_avl(arbre['gauche']) +
            parcours_postfixe_avl(arbre['droite']) +
            [arbre['valeur']])

```

## Exercice 4 — B-Arbre d'ordre m




### 1. Création de la structure

Écrire les fonctions `creer_noeud_b_arbre(feuille=True)` pour créer un nœud du B-arbre et `construire_b_arbre(valeurs, ordre)` pour construire un B-arbre à partir d’une liste de valeurs et de l’ordre donné.




```python
# Fonction : creer_noeud_b_arbre
# Rôle : crée un nœud de B-arbre.
def creer_noeud_b_arbre(feuille=True):
    """Crée un nœud de B-arbre."""
    return {"cles": [], "enfants": [], "feuille": feuille}


# Fonction : construire_b_arbre
# Rôle : construit un B-arbre à partir d'une liste de valeurs et de l'ordre donné.
def construire_b_arbre(valeurs, ordre):
    """Construit un B-arbre d'ordre `ordre` sans appeler `inserer_b_arbre`.

    Cette version construit une structure simple et hiérarchique :
    - les valeurs sont triées,
    - groupées par blocs de taille maximale (2*ordre - 1) pour former les feuilles,
    - puis les niveaux supérieurs sont construits progressivement jusqu'à la racine.
    """
    if not valeurs:
        return creer_noeud_b_arbre(feuille=True)

    max_cles = 2 * ordre - 1
    valeurs = sorted(set(valeurs))

    # Étape 1 : construire les feuilles
    niveau = []
    for i in range(0, len(valeurs), max_cles):
        noeud = creer_noeud_b_arbre(feuille=True)
        noeud["cles"] = valeurs[i:i + max_cles]
        niveau.append(noeud)

    # Étape 2 : remonter niveau par niveau jusqu'à la racine
    while len(niveau) > 1:
        nouveau_niveau = []

        # Un nœud interne de B-arbre avec k clés possède k+1 enfants
        max_enfants = 2 * ordre

        for i in range(0, len(niveau), max_enfants):
            enfants = niveau[i:i + max_enfants]
            parent = creer_noeud_b_arbre(feuille=False)
            parent["enfants"] = enfants

            # On prend comme séparateurs la première clé de chaque enfant à partir du 2e
            parent["cles"] = [enfant["cles"][0] for enfant in enfants[1:]]

            nouveau_niveau.append(parent)

        niveau = nouveau_niveau

    return niveau[0]
```

### 2. Propriétés du nœud

Écrire les fonctions `est_feuille(noeud)` et `est_plein(noeud, ordre)` permettant de vérifier respectivement si un nœud est une feuille et si un nœud est plein.


```python
# Fonction : est_feuille
# Rôle : vérifie si un nœud est une feuille.
def est_feuille(noeud):
    return noeud["feuille"]


# Fonction : est_plein
# Rôle : vérifie si un nœud est plein.
def est_plein(noeud, ordre):
    return len(noeud["cles"]) == 2 * ordre - 1
```

### 3. Opérations sur l’arbre

Écrire les fonctions `rechercher_b_arbre(arbre, cle)`, `inserer_b_arbre(arbre, cle, ordre)` et `supprimer_b_arbre(arbre, cle, ordre)` permettant respectivement de rechercher, d’insérer et de supprimer une clé dans le B-arbre.

Ainsi que les fonctions `diviser_noeud(parent, indice, ordre)`, `emprunter_frere_gauche(parent, indice)`, `emprunter_frere_droite(parent, indice)` et `fusionner_noeuds(parent, indice)` permettant de réorganiser l’arbre si nécessaire après une insertion ou une suppression.


```python
# Fonction : rechercher_b_arbre
# Rôle : recherche une clé dans un B-arbre.
def rechercher_b_arbre(arbre, cle):
    if arbre is None:
        return False

    i = 0
    while i < len(arbre["cles"]) and cle > arbre["cles"][i]:
        i += 1

    if i < len(arbre["cles"]) and cle == arbre["cles"][i]:
        return True

    if arbre["feuille"]:
        return False

    return rechercher_b_arbre(arbre["enfants"][i], cle)


# Fonction : diviser_noeud
# Rôle : divise un nœud plein lors d'une insertion.
def diviser_noeud(parent, indice, ordre):
    t = ordre
    enfant = parent["enfants"][indice]
    nouveau = creer_noeud_b_arbre(feuille=enfant["feuille"])

    cle_mediane = enfant["cles"][t - 1]

    nouveau["cles"] = enfant["cles"][t:]
    enfant["cles"] = enfant["cles"][:t - 1]

    if not enfant["feuille"]:
        nouveau["enfants"] = enfant["enfants"][t:]
        enfant["enfants"] = enfant["enfants"][:t]

    parent["cles"].insert(indice, cle_mediane)
    parent["enfants"].insert(indice + 1, nouveau)


# Fonction : _inserer_non_plein
# Rôle : insère une clé dans un nœud non plein.
def _inserer_non_plein(noeud, cle, ordre):
    i = len(noeud["cles"]) - 1

    if noeud["feuille"]:
        if cle in noeud["cles"]:
            return
        noeud["cles"].append(None)
        while i >= 0 and cle < noeud["cles"][i]:
            noeud["cles"][i + 1] = noeud["cles"][i]
            i -= 1
        if i >= 0 and noeud["cles"][i] == cle:
            noeud["cles"].pop()
            return
        noeud["cles"][i + 1] = cle
    else:
        while i >= 0 and cle < noeud["cles"][i]:
            i -= 1
        i += 1

        if est_plein(noeud["enfants"][i], ordre):
            diviser_noeud(noeud, i, ordre)
            if cle > noeud["cles"][i]:
                i += 1
            elif cle == noeud["cles"][i]:
                return

        _inserer_non_plein(noeud["enfants"][i], cle, ordre)


# Fonction : inserer_b_arbre
# Rôle : insère une clé dans un B-arbre.
def inserer_b_arbre(arbre, cle, ordre):
    if arbre is None:
        arbre = creer_noeud_b_arbre(feuille=True)

    racine = arbre
    if est_plein(racine, ordre):
        nouvelle_racine = creer_noeud_b_arbre(feuille=False)
        nouvelle_racine["enfants"].append(racine)
        diviser_noeud(nouvelle_racine, 0, ordre)
        racine = nouvelle_racine

    _inserer_non_plein(racine, cle, ordre)
    return racine
```

```python
# Fonction : emprunter_frere_gauche
# Rôle : emprunte une clé au frère gauche.
def emprunter_frere_gauche(parent, indice):
    enfant = parent["enfants"][indice]
    frere_gauche = parent["enfants"][indice - 1]

    enfant["cles"].insert(0, parent["cles"][indice - 1])
    parent["cles"][indice - 1] = frere_gauche["cles"].pop()

    if not frere_gauche["feuille"]:
        enfant["enfants"].insert(0, frere_gauche["enfants"].pop())


# Fonction : emprunter_frere_droite
# Rôle : emprunte une clé au frère droit.
def emprunter_frere_droite(parent, indice):
    enfant = parent["enfants"][indice]
    frere_droite = parent["enfants"][indice + 1]

    enfant["cles"].append(parent["cles"][indice])
    parent["cles"][indice] = frere_droite["cles"].pop(0)

    if not frere_droite["feuille"]:
        enfant["enfants"].append(frere_droite["enfants"].pop(0))


# Fonction : fusionner_noeuds
# Rôle : fusionne deux nœuds frères.
def fusionner_noeuds(parent, indice):
    enfant = parent["enfants"][indice]
    frere_droite = parent["enfants"][indice + 1]

    enfant["cles"].append(parent["cles"].pop(indice))
    enfant["cles"].extend(frere_droite["cles"])

    if not enfant["feuille"]:
        enfant["enfants"].extend(frere_droite["enfants"])

    parent["enfants"].pop(indice + 1)


# Fonction : _max_cle
# Rôle : retourne la clé maximale d'un sous-arbre.
def _max_cle(noeud):
    while not noeud["feuille"]:
        noeud = noeud["enfants"][-1]
    return noeud["cles"][-1]


# Fonction : _min_cle
# Rôle : retourne la clé minimale d'un sous-arbre.
def _min_cle(noeud):
    while not noeud["feuille"]:
        noeud = noeud["enfants"][0]
    return noeud["cles"][0]


# Fonction : _remplir_enfant
# Rôle : réorganise un enfant avant une suppression.
def _remplir_enfant(parent, indice, ordre):
    if indice > 0 and len(parent["enfants"][indice - 1]["cles"]) >= ordre:
        emprunter_frere_gauche(parent, indice)
    elif indice < len(parent["enfants"]) - 1 and len(parent["enfants"][indice + 1]["cles"]) >= ordre:
        emprunter_frere_droite(parent, indice)
    else:
        if indice < len(parent["enfants"]) - 1:
            fusionner_noeuds(parent, indice)
        else:
            fusionner_noeuds(parent, indice - 1)
            indice -= 1
    return indice


# Fonction : _supprimer_b_arbre
# Rôle : supprime une clé dans un B-arbre de manière récursive.
def _supprimer_b_arbre(noeud, cle, ordre):
    i = 0
    while i < len(noeud["cles"]) and cle > noeud["cles"][i]:
        i += 1

    if i < len(noeud["cles"]) and noeud["cles"][i] == cle:
        if noeud["feuille"]:
            noeud["cles"].pop(i)
            return

        if len(noeud["enfants"][i]["cles"]) >= ordre:
            pred = _max_cle(noeud["enfants"][i])
            noeud["cles"][i] = pred
            _supprimer_b_arbre(noeud["enfants"][i], pred, ordre)
        elif len(noeud["enfants"][i + 1]["cles"]) >= ordre:
            succ = _min_cle(noeud["enfants"][i + 1])
            noeud["cles"][i] = succ
            _supprimer_b_arbre(noeud["enfants"][i + 1], succ, ordre)
        else:
            fusionner_noeuds(noeud, i)
            _supprimer_b_arbre(noeud["enfants"][i], cle, ordre)
        return

    if noeud["feuille"]:
        return

    if len(noeud["enfants"][i]["cles"]) < ordre:
        i = _remplir_enfant(noeud, i, ordre)

    _supprimer_b_arbre(noeud["enfants"][i], cle, ordre)


# Fonction : supprimer_b_arbre
# Rôle : supprime une clé dans un B-arbre.
def supprimer_b_arbre(arbre, cle, ordre):
    if arbre is None:
        return None

    _supprimer_b_arbre(arbre, cle, ordre)

    if len(arbre["cles"]) == 0 and not arbre["feuille"]:
        arbre = arbre["enfants"][0]

    return arbre
```

### 4. Travail supplémentaire

Généraliser l’implémentation du B-arbre d’ordre 4 au cas d’un B-arbre d’ordre `m`.

Ici, toute l’implémentation est déjà généralisée grâce au paramètre `ordre`.


```python
# Fonction : parcours_largeur_b_arbre
# Rôle : effectue un parcours en largeur du B-arbre.
def parcours_largeur_b_arbre(arbre):
    if arbre is None:
        return []

    resultat = []
    file = [arbre]

    while file:
        noeud = file.pop(0)
        resultat.append(list(noeud["cles"]))
        if not noeud["feuille"]:
            file.extend(noeud["enfants"])

    return resultat
```
