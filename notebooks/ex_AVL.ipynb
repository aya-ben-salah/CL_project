<h3>Structure AVL<h3>

```python
class Noeud:
    def __init__(self, valeur):
        self.valeur = valeur
        self.gauche = None
        self.droite = None
        self.hauteur = 1
```

<h3>Fonctions utilitaires<h3>

```python
def hauteur(n):
    if n is None:
        return 0
    return n.hauteur


def facteur_equilibre(n):
    if n is None:
        return 0
    return hauteur(n.gauche) - hauteur(n.droite)


def maj_hauteur(n):
    n.hauteur = 1 + max(hauteur(n.gauche), hauteur(n.droite))
```

<h3>Rotation<h3>

```python
def rotation_droite(y):
    x = y.gauche
    T2 = x.droite

    x.droite = y
    y.gauche = T2

    maj_hauteur(y)
    maj_hauteur(x)

    return x


def rotation_gauche(x):
    y = x.droite
    T2 = y.gauche

    y.gauche = x
    x.droite = T2

    maj_hauteur(x)
    maj_hauteur(y)

    return y
```

<h3>Insertion AVL<h3>

```python
def inserer(avl, cle):
    if avl is None:
        return Noeud(cle)

    if cle < avl.valeur:
        avl.gauche = inserer(avl.gauche, cle)
    else:
        avl.droite = inserer(avl.droite, cle)

    maj_hauteur(avl)

    eq = facteur_equilibre(avl)

    # Cas gauche-gauche
    if eq > 1 and cle < avl.gauche.valeur:
        return rotation_droite(avl)

    # Cas droite-droite
    if eq < -1 and cle > avl.droite.valeur:
        return rotation_gauche(avl)

    # Cas gauche-droite
    if eq > 1 and cle > avl.gauche.valeur:
        avl.gauche = rotation_gauche(avl.gauche)
        return rotation_droite(avl)

    # Cas droite-gauche
    if eq < -1 and cle < avl.droite.valeur:
        avl.droite = rotation_droite(avl.droite)
        return rotation_gauche(avl)

    return avl
```

<h3>Recherche<h3>

```python
def rechercher(avl, cle):
    if avl is None:
        return False

    if cle == avl.valeur:
        return True
    elif cle < avl.valeur:
        return rechercher(avl.gauche, cle)
    else:
        return rechercher(avl.droite, cle)
```

<h3>Trouver minimum<h3>

```python
def min_valeur(noeud):
    while noeud.gauche is not None:
        noeud = noeud.gauche
    return noeud
```

<h3>Suppression AVL<h3>

```python
def supprimer(avl, cle):
    if avl is None:
        return avl

    if cle < avl.valeur:
        avl.gauche = supprimer(avl.gauche, cle)
    elif cle > avl.valeur:
        avl.droite = supprimer(avl.droite, cle)
    else:
        if avl.gauche is None:
            return avl.droite
        elif avl.droite is None:
            return avl.gauche

        temp = min_valeur(avl.droite)
        avl.valeur = temp.valeur
        avl.droite = supprimer(avl.droite, temp.valeur)

    maj_hauteur(avl)

    eq = facteur_equilibre(avl)

    # équilibrage
    if eq > 1 and facteur_equilibre(avl.gauche) >= 0:
        return rotation_droite(avl)

    if eq > 1 and facteur_equilibre(avl.gauche) < 0:
        avl.gauche = rotation_gauche(avl.gauche)
        return rotation_droite(avl)

    if eq < -1 and facteur_equilibre(avl.droite) <= 0:
        return rotation_gauche(avl)

    if eq < -1 and facteur_equilibre(avl.droite) > 0:
        avl.droite = rotation_droite(avl.droite)
        return rotation_gauche(avl)

    return avl
```

<h3>Parcours<h3>

```python
def parcours_infixe(arbre):
    if arbre:
        parcours_infixe(arbre.gauche)
        print(arbre.valeur, end=" ")
        parcours_infixe(arbre.droite)


from collections import deque

def parcours_largeur(arbre):
    if arbre is None:
        return

    file = deque([arbre])
    while file:
        n = file.popleft()
        print(n.valeur, end=" ")

        if n.gauche:
            file.append(n.gauche)
        if n.droite:
            file.append(n.droite)
```

<h3>Test<h3>

```python
avl = None

valeurs = [10, 20, 30, 40, 50, 25]

for v in valeurs:
    avl = inserer(avl, v)

print("Parcours infixe :", end=" ")
parcours_infixe(avl)

print("\nRecherche 25 :", rechercher(avl, 25))

avl = supprimer(avl, 40)

print("Après suppression :", end=" ")
parcours_infixe(avl)

print("\nParcours largeur :", end=" ")
parcours_largeur(avl)
```
