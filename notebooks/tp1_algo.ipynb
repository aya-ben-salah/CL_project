```python
# 1. Création de la structure
class Noeud:
    def __init__(self, valeur, gauche=None, droite=None):
        self.valeur = valeur
        self.gauche = gauche
        self.droite = droite


def creer_noeud(valeur, gauche=None, droite=None):
    return Noeud(valeur, gauche, droite)

def construire_arbre_binaire():
    n4 = creer_noeud(4)
    n5 = creer_noeud(5)
    n2 = creer_noeud(2, n4, n5)
    n3 = creer_noeud(3)
    racine = creer_noeud(1, n2, n3)
    return racine
```

```python
# 2. Propriétés de l’arbre


def taille_arbre(arbre):
    if arbre is None:
        return 0
    return 1 + taille_arbre(arbre.gauche) + taille_arbre(arbre.droite)


def hauteur_arbre(arbre):
    if arbre is None:
        return -1  # racine = niveau 0
    return 1 + max(hauteur_arbre(arbre.gauche), hauteur_arbre(arbre.droite))


def nombre_feuilles(arbre):
    if arbre is None:
        return 0
    if arbre.gauche is None and arbre.droite is None:
        return 1
    return nombre_feuilles(arbre.gauche) + nombre_feuilles(arbre.droite)


def nombre_noeuds_internes(arbre):
    if arbre is None or (arbre.gauche is None and arbre.droite is None):
        return 0
    return 1 + nombre_noeuds_internes(arbre.gauche) + nombre_noeuds_internes(arbre.droite)
```

```python
# 3. Parcours de l’arbre


def parcours_prefixe(arbre):
    if arbre is not None:
        print(arbre.valeur, end=" ")
        parcours_prefixe(arbre.gauche)
        parcours_prefixe(arbre.droite)


def parcours_infixe(arbre):
    if arbre is not None:
        parcours_infixe(arbre.gauche)
        print(arbre.valeur, end=" ")
        parcours_infixe(arbre.droite)


def parcours_postfixe(arbre):
    if arbre is not None:
        parcours_postfixe(arbre.gauche)
        parcours_postfixe(arbre.droite)
        print(arbre.valeur, end=" ")


from collections import deque

def parcours_largeur(arbre):
    if arbre is None:
        return

    file = deque([arbre])

    while file:
        noeud = file.popleft()
        print(noeud.valeur, end=" ")

        if noeud.gauche:
            file.append(noeud.gauche)
        if noeud.droite:
            file.append(noeud.droite)
```

```python
arbre = construire_arbre_binaire()

print("Taille :", taille_arbre(arbre))
print("Hauteur :", hauteur_arbre(arbre))
print("Nombre de feuilles :", nombre_feuilles(arbre))
print("Nombre de noeuds internes :", nombre_noeuds_internes(arbre))

print("Parcours préfixe :", end=" ")
parcours_prefixe(arbre)

print("\nParcours infixe :", end=" ")
parcours_infixe(arbre)

print("\nParcours postfixe :", end=" ")
parcours_postfixe(arbre)

print("\nParcours en largeur :", end=" ")
parcours_largeur(arbre)
```

```python
# Structure
class Noeud:
    def __init__(self, valeur, gauche=None, droite=None):
        self.valeur = valeur
        self.gauche = gauche
        self.droite = droite


def creer_noeud(valeur, gauche=None, droite=None):
    return Noeud(valeur, gauche, droite)
```

```python
# 1. Construction ABR

def inserer_abr(arbre, cle):
    if arbre is None:
        return creer_noeud(cle)

    if cle < arbre.valeur:
        arbre.gauche = inserer_abr(arbre.gauche, cle)
    elif cle > arbre.valeur:
        arbre.droite = inserer_abr(arbre.droite, cle)

    return arbre


def construire_abr(valeurs):
    arbre = None
    for v in valeurs:
        arbre = inserer_abr(arbre, v)
    return arbre
```

```python
# 2. Propriétés (réutilisées)


def taille_arbre(arbre):
    if arbre is None:
        return 0
    return 1 + taille_arbre(arbre.gauche) + taille_arbre(arbre.droite)


def hauteur_arbre(arbre):
    if arbre is None:
        return -1
    return 1 + max(hauteur_arbre(arbre.gauche), hauteur_arbre(arbre.droite))


def nombre_feuilles(arbre):
    if arbre is None:
        return 0
    if arbre.gauche is None and arbre.droite is None:
        return 1
    return nombre_feuilles(arbre.gauche) + nombre_feuilles(arbre.droite)


def nombre_noeuds_internes(arbre):
    if arbre is None or (arbre.gauche is None and arbre.droite is None):
        return 0
    return 1 + nombre_noeuds_internes(arbre.gauche) + nombre_noeuds_internes(arbre.droite)
```

```python
#parcoure
def parcours_prefixe(arbre):
    if arbre:
        print(arbre.valeur, end=" ")
        parcours_prefixe(arbre.gauche)
        parcours_prefixe(arbre.droite)


def parcours_infixe(arbre):
    if arbre:
        parcours_infixe(arbre.gauche)
        print(arbre.valeur, end=" ")
        parcours_infixe(arbre.droite)


def parcours_postfixe(arbre):
    if arbre:
        parcours_postfixe(arbre.gauche)
        parcours_postfixe(arbre.droite)
        print(arbre.valeur, end=" ")


from collections import deque

def parcours_largeur(arbre):
    if arbre is None:
        return

    file = deque([arbre])

    while file:
        noeud = file.popleft()
        print(noeud.valeur, end=" ")

        if noeud.gauche:
            file.append(noeud.gauche)
        if noeud.droite:
            file.append(noeud.droite)
```

```python
# 3. Opérations ABR
def rechercher_abr(arbre, cle):
    if arbre is None:
        return False

    if cle == arbre.valeur:
        return True
    elif cle < arbre.valeur:
        return rechercher_abr(arbre.gauche, cle)
    else:
        return rechercher_abr(arbre.droite, cle)


def trouver_min(arbre):
    while arbre.gauche is not None:
        arbre = arbre.gauche
    return arbre


def supprimer_abr(arbre, cle):
    if arbre is None:
        return None

    if cle < arbre.valeur:
        arbre.gauche = supprimer_abr(arbre.gauche, cle)

    elif cle > arbre.valeur:
        arbre.droite = supprimer_abr(arbre.droite, cle)

    else:
        if arbre.gauche is None and arbre.droite is None:
            return None

        if arbre.gauche is None:
            return arbre.droite

        if arbre.droite is None:
            return arbre.gauche

        temp = trouver_min(arbre.droite)
        arbre.valeur = temp.valeur
        arbre.droite = supprimer_abr(arbre.droite, temp.valeur)

    return arbre
```

```python
# TEST

valeurs = [50, 30, 70, 20, 40, 60, 80]
arbre = construire_abr(valeurs)

print("Taille :", taille_arbre(arbre))
print("Hauteur :", hauteur_arbre(arbre))
print("Feuilles :", nombre_feuilles(arbre))
print("Noeuds internes :", nombre_noeuds_internes(arbre))

print("Prefixe :", end=" ")
parcours_prefixe(arbre)

print("\nInfixe :", end=" ")
parcours_infixe(arbre)

print("\nPostfixe :", end=" ")
parcours_postfixe(arbre)

print("\nLargeur :", end=" ")
parcours_largeur(arbre)

print("\nRecherche 40 :", rechercher_abr(arbre, 40))

arbre = supprimer_abr(arbre, 30)

print("Apres suppression :", end=" ")
parcours_infixe(arbre)
```
